```python
df_copy=df.copy()
```

```python
# deleting the non numeric column
df_copy=df_copy.drop(df_copy.index[10472])
```

```python
#making sure there are no non-numeric values present
df_copy[~df_copy['Reviews'].str.isnumeric()]
```

```python
# converting the reviews column from object to integer
df_copy['Reviews']=df_copy['Reviews'].astype(int)
```

```python
df_copy['Size'].unique()
```

```python
# checking to see if the size column has any missing values
df_copy['Size'].isnull().sum()
```

```python
#Making the data in size column easy to read and converting to suitable datatype
df_copy['Size'] = df_copy['Size'].astype(str)
df_copy['Size']=df_copy['Size'].str.replace('M','000')
df_copy['Size']=df_copy['Size'].str.replace('k','')
df_copy['Size']=df_copy['Size'].replace('Varies with device',np.nan)
df_copy['Size']=df_copy['Size'].astype(float)
```

```python
df_copy['Size']
```

```python
#dealing with other columns to clean the data
df['Installs'].unique()
```

```python
df['Price'].unique()
```

```python
#Removing any extra characters from the data.
chars_to_remove=['+',',','$']
cols_to_clean=['Installs','Price']
for col in cols_to_clean:
  for item in chars_to_remove:
    df_copy[col]=df_copy[col].str.replace(item,'')
```

```python
df_copy['Price'].unique()
```

```python
# coverting the datatype from object to integer
df_copy['Installs']=df_copy['Installs'].astype(int)
```

```python
##last update feature
df_copy['Last Updated'].unique()
```

```python
# extracting day,month and year
df_copy['Last Updated']=pd.to_datetime(df_copy['Last Updated'])
df_copy['Day']=df_copy['Last Updated'].dt.day
df_copy['Month']=df_copy['Last Updated'].dt.month
df_copy['Year']=df_copy['Last Updated'].dt.year
```

```python
df_copy.to_csv('cleaned_googleplaystore.csv', index=False)
```

```python
##TOP 10 APP CATEGORIES
category=pd.DataFrame(df_copy['Category'].value_counts())
category.rename(columns={'Category':'Count'},inplace=True)
```

```python
plt.figure(figsize=(15,6))
sns.barplot(x=category.index[:10],y='count',data=category[:10],palette='hls')
plt.title('Top 10 app categories')
plt.xticks(rotation=90)
plt.xlabel('Category')
plt.ylabel('Count')
plt.show()
```

```python
df_cat_installs = df_copy.groupby(['Category'])['Installs'].sum().sort_values(ascending = False).reset_index()
df_cat_installs.Installs = df_cat_installs.Installs/1000000000# converting into billions
df2 = df_cat_installs.head(10)
plt.figure(figsize = (14,10))
sns.set_context("talk")
sns.set_style("darkgrid")

ax = sns.barplot(x = 'Installs' , y = 'Category' , data = df2 )
ax.set_xlabel('No. of Installations in Billions')
ax.set_ylabel('')
ax.set_title("Most Popular Categories in Play Store", size = 20)
```

```python
## shows the most popular category in every category
dfa = df_copy.groupby(['Category' ,'App'])['Installs'].sum().reset_index()
dfa = dfa.sort_values('Installs', ascending = False)
apps = ['GAME', 'COMMUNICATION', 'PRODUCTIVITY', 'SOCIAL' ]
sns.set_context("poster")
sns.set_style("darkgrid")

plt.figure(figsize=(40,30))

for i,app in enumerate(apps):
    df2 = dfa[dfa.Category == app]
    df3 = df2.head(5)
    plt.subplot(4,2,i+1)
    sns.barplot(data= df3,x= 'Installs' ,y='App' )
    plt.xlabel('Installation in Millions')
    plt.ylabel('')
    plt.title(app,size = 20)

plt.tight_layout()
plt.subplots_adjust(hspace= .3)
plt.show()
```
